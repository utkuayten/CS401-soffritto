```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import  seaborn as sns
import sys
import os

project_root = os.path.abspath(os.path.join(os.getcwd(), "..")) # Add project root (parent of "visualize/") to sys.path
sys.path.append(project_root)

from results.genomic_multitarget_informer.metrics_calculate import evaluate_all_metrics

y_true = np.load("../results/genomic_multitarget_informer/true.npy")
y_pred = np.load("../results/genomic_multitarget_informer/pred.npy")

metrics = evaluate_all_metrics(y_true, y_pred)
print(metrics)

metrics_to_plot = ['ARFE', 'KL', 'Spearman', 'KS', 'Wasserstein']
means = {
    'ARFE': metrics['Mean ARFE'],
    'KL': metrics['Mean KL'],
    'Spearman': metrics['Mean Spearman'],
    'KS': metrics['Mean KS'],
    'Wasserstein': metrics['Mean Wasserstein']
}



save_dir = "../visualize/plots"
os.makedirs(save_dir, exist_ok=True)

for metric in metrics_to_plot: # Plot each metric in its own figure
    plt.figure(figsize=(4, 8))
    sns.violinplot(y=metrics[metric], inner='box')
    plt.plot([-0.4, 0.4], [means[metric], means[metric]], color='red', linestyle='--', linewidth=2)
    plt.title(f"{metric} Distribution with Mean Line")
    plt.ylabel("Value")
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plot_filename = os.path.join(save_dir, f"{metric}_violin_plot.png")
    plt.savefig(plot_filename)
    plt.close()  # Close the figure to free memory
    plt.show()
```
